## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
# TULIS KODE ANDA DI SINI (Latihan 1)
import seaborn as sns
import matplotlib.pyplot as plt, seaborn as sns
df = sns.load_dataset('tips')
print(df.head()) # Menampilkan 5 data pertama
print(df.shape) # Menampilkan jumlah baris dan kolom
df.info()# Menampilkan informasi dataset

```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
# TULIS KODE ANDA DI SINI (Latihan 2)
df['total_bill'].hist(figsize=(9,6), bins=15)
plt.title('Histogram Total Bill')
plt.show()
sns.boxplot(data=df[['tip']])
plt.title('Deteksi Outlier Tip')
plt.show()

```

Sebaran total_bill lebih banyak berada pada nilai rendah hingga menengah. Pada box plot tip terlihat beberapa titik di luar whisker yang menunjukkan adanya outlier atau nilai tip yang cukup jauh dari data lainnya.

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# TULIS KODE ANDA DI SINI (Latihan 3)
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time')
plt.title('Total Bill vs Tip berdasarkan Time')
plt.show()
```

Dapat disimpulkan bahwa terdapat hubungan positif antara total_bill dan tip. Semakin besar nilai total tagihan, cenderung semakin besar juga nilai tip. Baik pada waktu Lunch maupun Dinner terlihat pola yang sama.

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
# TULIS KODE ANDA DI SINI (Latihan 4)

num = df.select_dtypes(include='number')
sns.heatmap(num.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()

```

Pasangan fitur yang paling berkorelasi adalah total_bill dan tip dengan korelasi sekitar 0,68. Korelasinya positif, artinya semakin besar total tagihan, cenderung semakin besar juga tip yang diberikan.
