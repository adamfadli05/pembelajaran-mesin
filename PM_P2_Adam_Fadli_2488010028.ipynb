# 1.	Apa perbedaan supervised, unsupervised, dan reinforcement learning?



*   Supervised : Pembelajaran data yang sudah ada labelnya, Model belajar dari memetakan fitur dan jawaban untuk melakukan prediksi.
*   unsuvervised : menggunakan data yang tidak ada label nya, Model mencari pola atau struktur tersembunyi dalam data.
*   reinforcement : pembebelajaran dari interaksi via reward & funishment






# 2.	Mengapa pembagian data latih dan uji penting?

Pembagian data latih dan uji penting untuk mengetahui apakah model mampu memberikan prediksi yang baik pada data yang belum pernah dilihat sebelumnya. Data latih digunakan untuk mengajari model sebanyak 80%, sedangkan data uji digunakan untuk mengukur kemampuan model sebanyak 20%. Dengan demikian, data uji harus dipisahkan dari proses pelatihan.

# 3.	Mengapa alur kerja ML disebut iteratif?

Alur kerja Machine Learning disebut iteratif karena prosesnya dapat dilakukan berulang kali. Jika hasil evaluasi model belum baik, kita dapat kembali memperbaiki data, melakukan preprocessing, memilih fitur yang lebih sesuai, mengganti algoritma, atau melakukan tuning.
